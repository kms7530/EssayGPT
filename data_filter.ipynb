```python
import os
import json
```

```python
def get_file_list(folder_path):
    file_list = []
    for root, dirs, files in os.walk(folder_path):
        for file in files:
            file_list.append(os.path.join(root, file))
    return file_list
```

```python
folder_path = './data'
all_files = get_file_list(folder_path)
```

```python
filter_in = ["고등"]
filter_out = ["설명글", "filtered"]
cnt = 0

for file in all_files:
    is_pass = True
    contents = {}

    if "json" not in file:
        continue

    for filter_ in filter_in:
        if filter_ not in file:
            is_pass = False
            break

    for filter_ in filter_out:
        if filter_ in file:
            is_pass = False
            break

    if not is_pass:
        continue

    print(file)
    cnt += 1
    continue
    with open(file, 'r', encoding='utf-8') as json_file:
        json_data = json.load(json_file)

        contents['essay_id'] = json_data['info']['essay_id']
        contents['essay_type'] = json_data['info']['essay_type']
        contents['essay_main_subject'] = json_data['info']['essay_main_subject']
        contents['essay_sub_subject'] = " ".join([para["paragraph_txt"] for para in json_data['paragraph']]).replace("#@문장구분#", "")

        with open(f"./filtered/{contents['essay_id']}.json", 'w', encoding='utf-8') as f:
            json.dump(contents, f, ensure_ascii=False)

print(cnt)
```
